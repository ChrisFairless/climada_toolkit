# OpenStreetMap critical infrastructure

Load healthcare features for Liechtenstein using the toolkit's thin wrapper around
[osm-flex](https://github.com/osm-flex/osm-flex). Change `ci_type` to select another
preset: `education`, `healthcare`, `water`, `telecom`, `road`, `rail`, `air`,
`gas`, `oil`, `power`, `wastewater` or `food`.

The wrapper downloads a Geofabrik PBF, extracts the selected infrastructure and
returns CLIMADA `Exposures`, retaining the OSM geometries and tags. Large-country
downloads can be substantial. Some extracts group countries or cover only part
of a country; no additional country clipping is applied.

Data: © [OpenStreetMap contributors](https://www.openstreetmap.org/copyright),
available under the ODbL and distributed by [Geofabrik](https://download.geofabrik.de/).
OSM completeness varies, and a facility may be mapped as both a point and a polygon.


````python
import logging

from climada_toolkit.inputs.exposures.openstreetmap import openstreetmap

logging.basicConfig(level=logging.INFO)
COUNTRY = "LIE"
CI_TYPE = "road"

exp = openstreetmap.load_osm_critical_infra(COUNTRY, CI_TYPE)
````

The wrapper stores downloads and HDF5 outputs under CLIMADA's local data directory.
It automatically saves `osm_{country}_{ci_type}.hdf5` and reuses that output on
subsequent calls. Set `overwrite=True` to refresh both the raw download and the
processed file.


````python
import contextily as ctx

# Plot original geometries, including points and polygons.
ax = exp.gdf.plot(figsize=(8, 8))
ctx.add_basemap(ax, zoom=12, source=ctx.providers.Esri.WorldTopoMap)
ax.set_title(f"OSM {CI_TYPE}: {COUNTRY} extract")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
````

The wrapper retrieves features; it does not assign exposure values automatically. Before impact calculations, choose values and units appropriate to your analysis.

Lines and polygons require CLIMADA's [`lines_polys_handler`](https://climada-python.readthedocs.io/en/stable/user-guide/climada_entity_Exposures_polygons_lines.html) tools.

See the adjacent [README](README.md) for data-quality notes and links to osm-flex's custom queries and simplification tools.

## Optional cleanup

These calls are examples only, so running the notebook retains its data.
They remove matching files directly inside the selected directories.

```python
from climada_toolkit.inputs.exposures.openstreetmap import openstreetmap

# Remove raw *.osm.pbf downloads:
openstreetmap.clear_osm_downloads()

# Remove saved *.hdf5 exposures:
openstreetmap.clear_osm_outputs()

# Or remove both:
openstreetmap.clear_all_osm_files()
```
