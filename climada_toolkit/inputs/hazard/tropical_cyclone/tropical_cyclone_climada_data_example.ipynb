## Historical TC events

This loads data from the CLIMADA Data API. Note that the wind footprints are old, made with an algorithm where bugs have since been fixed, and with very noisy footprints, since the tracks were insufficiently interpolated.

However this is consistent with the future sets, and so for a quick analysis you kind of have to use them.

To work with individual events we recommend creating your own footprints with CLIMADA's internal tools. See the tutorial [here]().


```python
from climada_toolkit.inputs.hazard.tropical_cyclone.tropical_cyclone import (
    load_historic_tropical_cyclone,
)

hazard = load_historic_tropical_cyclone(country="DOM")

print(f"Loaded {hazard.size} tropical cyclone events")

hazard.plot_intensity(0)
```
